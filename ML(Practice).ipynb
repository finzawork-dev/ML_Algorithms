```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=500, n_features=1, noise=15.0, random_state=42)  #this function makes a custom dataset
```

```python
df = pd.DataFrame(X, columns=["Study hours"])
df["Exam score"]=y

df_clean = df[(df["Study hours"]>=0) & (df["Exam score"] >=0 )]
print(df_clean.head())

df_clean.to_csv("LinearRegression_practice.csv", index=False)
```
